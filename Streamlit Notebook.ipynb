# Data Preparation for Streamlit
The clean dataset from the analysis notebook is loaded, checked and saved in a format that keeps the data types, so the Streamlit app can use it directly.

```python
import pandas as pd
```

## Load the Clean CSV File

```python
df = pd.read_csv('finance_banking_transactions_clean.csv')
df.info()
```

**Observation:** The file has 52,000 rows and 23 columns, but `transaction_date` is read as text (`str`) instead of `datetime`, because a CSV file does not store data types.

## Fix the Date Type

```python
df['transaction_date'] = pd.to_datetime(df['transaction_date'])
df['transaction_date'].dtype
```

## Save as Parquet
Parquet keeps the data types (e.g. `transaction_date` stays `datetime`) and is about 3.5 times smaller than CSV.

```python
df.to_parquet('finance_banking_transactions_clean.parquet', index=False)

# Check the saved file
df_parquet = pd.read_parquet('finance_banking_transactions_clean.parquet')
df_parquet.info()
```

**Result:** The Parquet file keeps all 52,000 rows and 23 columns, and `transaction_date` is saved as `datetime`. The Streamlit app reads this file.
